```python
import pandas as pd
df = pd.read_csv("netflix_movies.csv")
print("=== Initial Data Summary ===")
df.info()
```

```python
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
print("\n=== Cleaned Column Headers ===")
print(df.columns.tolist())
```

```python
initial_rows = len(df)
df = df.drop_duplicates()
print(f"\nDuplicates Inremoved: {initial_rows - len(df)}")
```

```python
print("\n=== Missing Values Count===")
print(df.isnull().sum())
```

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")

df = df.dropna(subset=["date_added", "rating", "duration"])

print("\n=== Missing Values Count (After Treatment) ===")
print(df.isnull().sum())
```

```python
text_columns = ["type", "title", "country", "rating"]
for col in text_columns:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"].str.strip(), errors="coerce")
```

```python
print("\n=== Final Dataset Overview ===")
print(df.info())
```

```python
print("\n=== First 5 Rows of Cleaned Data ===")
display(df.head())
```

```python
df.to_csv("cleaned_netflix_movies.csv", index=False)
print("\nCleaned dataset saved successfully as 'cleaned_netflix_movies.csv'!")
```
